```python
import os
import numpy as np
import matplotlib
matplotlib.use('Agg')  # Grafik arka planı belirtiyoruz (önceki kod için)
import matplotlib.pyplot as plt

# Dünya boyutları ve rüzgar durumu
WORLD_HEIGHT = 7
WORLD_WIDTH = 10
WIND = [0, 0, 0, 1, 1, 1, 2, 2, 1, 0]
START = [3, 0]
GOAL = [3, 7]
REWARD = -1.0
EPSILON = 0.1
ALPHA = 0.5

# Aksiyonlar
ACTION_UP = 0
ACTION_DOWN = 1
ACTION_LEFT = 2
ACTION_RIGHT = 3
ACTIONS = [ACTION_UP, ACTION_DOWN, ACTION_LEFT, ACTION_RIGHT]

def step(state, action):
    i, j = state
    if action == ACTION_UP:
        next_state = [max(i - 1 - WIND[j], 0), j]
    elif action == ACTION_DOWN:
        next_state = [max(min(i + 1 - WIND[j], WORLD_HEIGHT - 1), 0), j]
    elif action == ACTION_LEFT:
        next_state = [max(i - WIND[j], 0), max(j - 1, 0)]
    elif action == ACTION_RIGHT:
        next_state = [max(i - WIND[j], 0), min(j + 1, WORLD_WIDTH - 1)]
    else:
        raise ValueError("Invalid action")
    return next_state

def episode(q_value):
    state = START
    time = 0
    if np.random.binomial(1, EPSILON) == 1:
        action = np.random.choice(ACTIONS)
    else:
        action = np.argmax(q_value[state[0], state[1], :])
    while state != GOAL:
        next_state = step(state, action)
        if np.random.binomial(1, EPSILON) == 1:
            next_action = np.random.choice(ACTIONS)
        else:
            next_action = np.argmax(q_value[next_state[0], next_state[1], :])
        q_value[state[0], state[1], action] += ALPHA * (REWARD + q_value[next_state[0], next_state[1], next_action] - q_value[state[0], state[1], action])
        state = next_state
        action = next_action
        time += 1
    return time

def figure_6_3():
    q_value = np.zeros((WORLD_HEIGHT, WORLD_WIDTH, 4))
    episode_limit = 500
    steps = []
    for ep in range(episode_limit):
        steps.append(episode(q_value))
    steps = np.add.accumulate(steps)

    plt.plot(steps, np.arange(1, len(steps) + 1))
    plt.xlabel('Time steps')
    plt.ylabel('Episodes')
    plt.title('SARSA Learning Curve')

    save_path = os.path.join(r'C:\Users\mehat\Desktop\Anaconda Prompt\REİNFORCEMENT LEARNING\pr6 final project', 'figure_6_3.png')
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    plt.savefig(save_path)
    plt.close()

    optimal_policy = [['' for _ in range(WORLD_WIDTH)] for _ in range(WORLD_HEIGHT)]
    for i in range(WORLD_HEIGHT):
        for j in range(WORLD_WIDTH):
            if [i, j] == GOAL:
                optimal_policy[i][j] = 'G'
            else:
                bestAction = np.argmax(q_value[i, j, :])
                if bestAction == ACTION_UP:
                    optimal_policy[i][j] = 'U'
                elif bestAction == ACTION_DOWN:
                    optimal_policy[i][j] = 'D'
                elif bestAction == ACTION_LEFT:
                    optimal_policy[i][j] = 'L'
                elif bestAction == ACTION_RIGHT:
                    optimal_policy[i][j] = 'R'

    print('Optimal policy is:')
    for row in optimal_policy:
        print(row)
    print('Wind strength for each column:\n{}'.format([str(w) for w in WIND]))

if __name__ == '__main__':
    figure_6_3()
```
